## Optional reproduction

Set RUN_INFERENCE=True only after installing the dependencies documented in README.txt and selecting a CUDA GPU with BF16 support. This downloads the pinned public models and runs only this assignment's experiments. It may consume paid compute. New outputs use a separate directory; the saved evidence is preserved.


```python
RUN_INFERENCE=False
if RUN_INFERENCE:
    import shutil
    rerun=root/'reproduction'
    rerun.mkdir(exist_ok=True)
    for p in root.glob('*.py'):shutil.copy2(p,rerun/p.name)
    shutil.copytree(root/'data',rerun/'data',dirs_exist_ok=True)
    for phase in ['faithfulness','interventions']:
        subprocess.run([sys.executable,'experiments.py','--phase',phase,'--batch-size','16'],cwd=rerun,check=True)
    print('Fresh inference outputs saved in',rerun/'results')
else:
    print('Saved evidence verified; no inference, model download or training requested.')
```
